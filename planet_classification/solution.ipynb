### Context

**Planet Classes**
* **Class 0:** Desert
* **Class 1:** Volcanic
* **Class 2:** Oceanic
* **Class 3:** Forest
* **Class 4:** Icy

We must create a model that learns the pattern of a planet's characteristics (such as atmospheric pressure, average temperature, etc.) and is able to classify it among the classes above based solely on these characteristics.

We receive **training** data, which already comes with *labels* (the "answer key"), and **test** data, which has no answer key and is what we have to classify and submit.

### Loading the data

```python
import pandas as pd

train = pd.read_csv('./data/train.csv')
test = pd.read_csv('./data/test.csv')
```

```python
train
```

```python
test
```

### Analyzing the data

Let's run `describe()`, which is a DataFrame method to see general information about the data.

```python
train.describe()
```

Notice that the minimums and maximums of each column are completely unbalanced, since they are measured on different scales (almost like Celsius and Fahrenheit). 

Some have a range of [2, 71], while others have [0.01, 9], for example.

For this, we use the `StandardScaler`, responsible for making all data proportional.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
train_scaled = scaler.fit_transform(train.drop('target', axis=1))
test_scaled = scaler.transform(test)
targets = train['target']

pd.DataFrame(train_scaled).describe()
```

Now it gets a little harder to see because of the scientific notation, but notice that all minimums are around -4 and all maximums around 4. Furthermore, the standard deviations, `std`, are all exactly equal to 1.

### Modeling and training

With the data analysis done, we can start training.

We start by creating a split of training and validation data, so we can know how well the training is going while we train.

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(
    train, targets, test_size=0.2, random_state=42
)
```

`X_train`: training data

`y_train`: training *labels*

`X_val`: validation data

`y_val`: validation *labels*

Notice that `X` is capitalized and `y` is lowercase. This happens because X is a vector (all columns of each row), and y is a number (the `target` of that row). When there is a matrix/vector, it is convention to name it in uppercase.

Now, we take our `StandardScaler` from before, and join it with our classifier in a pipeline.

Here, I chose the **MLPClassifier** (a Neural Network), but it could have been any other; this is a matter of trial and error combined with preference.

```python
from sklearn.pipeline import Pipeline
from sklearn.neural_network import MLPClassifier

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('classifier', MLPClassifier(max_iter=1000, random_state=42))
])

pipeline
```

We train the pipeline

```python
pipeline.fit(X_train, y_train)
```

And we generate our predictions on the validation set

```python
val_predictions = pipeline.predict(X_val)
```

### Validating the model

Finally, we just check what **F1 Score** our model received.

```python
from sklearn.metrics import f1_score

print(f'F1 Score de {f1_score(y_val, val_predictions, average="weighted")}')
```

In this example, we ended up with **~80%** F1 score on the validation data, which would be more than enough to pass the phase in 2025.

It is important to reiterate that this is just ONE example of a solution and data analysis, among many possible ones. There were different possible analyses, more powerful models that can train for hours, or simply a little more luck (since there is a bit of randomness involved), which could have given us a better result on the test data, but the main idea here was to give a good sense of how to approach the problem.

### Submitting the answer
With everything ready, we just need to pass our data to a `.csv` file and the issue is resolved.

```python
test_predictions = pipeline.predict(test_scaled)
final_dataframe = pd.DataFrame({
    'id': test['id'],
    'target': test_predictions
})

final_dataframe
```

```python
final_dataframe.to_csv('submission.csv', index=False)
```
